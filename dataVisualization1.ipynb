```python
# Imports
import seaborn as sns#data visualization
import matplotlib.pyplot as plt
import pandas as pd

# Set style Changes the design/style of graphs.
sns.set(style="darkgrid")

# Load dataset
titanic = sns.load_dataset("titanic")
```

```python
# Check dataset
print(titanic.head())#1st 5 rows
print(titanic.tail())#last 5 
print(titanic.info())
print(titanic.describe(include="all"))#Shows statistical summary.
print("Shape:", titanic.shape)
print("Size:", titanic.size)
print("Dimensions:", titanic.ndim)
print("Columns:", titanic.columns)
```

```python
# 1. Distribution Plot - Age
plt.figure()#Creates new blank graph figure.
sns.histplot(titanic['age'].dropna(), kde=True)#Creates histogram.
plt.title("Distribution Plot - Passenger Age")
plt.show()#Most passengers were between 20–40 years.
```

```python
# 3.2 The Joint Plot - Age vs Fare
sns.jointplot(x='age', y='fare', data=titanic, kind='scatter', color='purple')
plt.suptitle("Joint Plot - Age vs Fare", y=1.02)
plt.show()
```

```python
# 3.3 The Pair Plot - Age, Fare, and Pclass (with survival hue) Creates multiple graphs together.Color changes based on survival.
sns.pairplot(titanic[['age', 'fare', 'pclass', 'survived']].dropna(), hue='survived', palette='Set1')
plt.suptitle("Pair Plot - Age, Fare, Pclass by Survival", y=1.02)
plt.show()
```

```python
# 3.4 The Rug Plot - Fare Shows small vertical lines for each fare value.
plt.figure(figsize=(6, 4))
sns.rugplot(titanic['fare'].dropna(), height=0.2)
plt.title("Rug Plot - Fare")
plt.show()#Useful for seeing dense regions.
```

```python
# 4. Categorical Plots

# 4.1 The Bar Plot - Average Fare by Class
plt.figure(figsize=(6, 4))
sns.barplot(x='class', y='fare', data=titanic)
plt.title("Bar Plot - Average Fare by Class")
plt.show()#Average fare for each class.
```

```python
# 4.2 The Count Plot - Number of Passengers by Embarkation Port
plt.figure(figsize=(6, 4))#Counts passengers from each embarkation port.
sns.countplot(x='embarked', data=titanic, palette='Set2')
plt.title("Count Plot - Passengers by Embarkation Port")
plt.show()##Which port had most passengers.
```

```python
# 4.3 The Box Plot - Age by Class Shows age distribution by class.
plt.figure(figsize=(6, 4))
sns.boxplot(x='class', y='age', data=titanic, palette='coolwarm')
plt.title("Box Plot - Age Distribution by Class")
plt.show()#Which class had older passengers? 1st
```

```python
# 4.4 The Violin Plot - Age by Sex Shape looks like violin.
plt.figure(figsize=(6, 4))
sns.violinplot(x='sex', y='age', data=titanic, palette='muted')
plt.title("Violin Plot - Age by Sex")
plt.show()#Shows age distribution for males and females.
```

```python
# 4.5 The Strip Plot - Fare by Class
plt.figure(figsize=(6, 4))
sns.stripplot(x='class', y='fare', data=titanic, jitter=True, color='orange')
plt.title("Strip Plot - Fare by Class")
plt.show()#Shows individual data points.See fare variation in each class.
```

```python
# 4.6 The Swarm Plot - Age by Class Like strip plot but automatically arranges points neatly.
plt.figure(figsize=(6, 4))
sns.swarmplot(x='class', y='age', data=titanic, palette='Set1')
plt.title("Swarm Plot - Age by Class")#Third class has many crowded points
#First class has fewer but older passengers
plt.show()#Individual passenger ages for each class
```

```python
# 5. Combining Swarm and Violin Plots - Age by Class
plt.figure(figsize=(6, 4))
sns.violinplot(x='class', y='age', data=titanic, inner=None, palette='pastel')
sns.swarmplot(x='class', y='age', data=titanic, color='k', alpha=0.6)
plt.title("Combined Violin + Swarm Plot - Age by Class")
plt.show()
```
