```python
# Group by species and describe sepal length Groups rows based on species.
summary = df.groupby('Species')['SepalLengthCm'].describe()

print("\nSummary Statistics:")
print(summary)
```

```python
# Custom statistics function
def custom_describe(x):
    mode_value = st.mode(x)#Most repeated value.
    median_value = np.median(x) #Middle value after sorting.
    return pd.Series(#Returns multiple values as Series.
        [x.count(), x.mean(), x.std(),
         mode_value, median_value,
         x.min(), x.max()],
        index=['count', 'mean', 'std',
               'mode', 'median',
               'min', 'max']
    )
```

```python
# Apply custom function
summary_with_mode_median = df.groupby('Species')['SepalLengthCm'].apply(custom_describe)
#Group data by species
#Send each group to custom function
#Function calculates statistics
#Results combined together
print("\nCustom Statistics:")
print(summary_with_mode_median)
```

```python
# Mode
mode_summary = df.groupby('Species')['SepalLengthCm'].apply(lambda x: st.mode(x))
#Small anonymous function.
print("\nMode:")
print(mode_summary)
```

```python

# Median
median_summary = df.groupby('Species')['SepalLengthCm'].median()

print("\nMedian:")
print(median_summary)
```

```python

# Mean
mean_summary = df.groupby('Species')['SepalLengthCm'].mean()

print("\nMean:")
print(mean_summary)
```

```python
# Numeric values for each species
species_numeric_value = df.groupby('Species').size().tolist()
#Convert Group Sizes to List
print("\nNumeric Values:")
print(species_numeric_value)
```

```python
# Full statistics for all species
species_stats = df.groupby('Species').describe()
#Statistics for ALL numeric columns grouped by species.
print("\nSpecies Statistics:")
print(species_stats)
```
