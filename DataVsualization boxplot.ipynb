```python
# Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns#Better visualization than plain matplotlib
```

```python
# Set style
sns.set(style="darkgrid")#Sets graph background style.
```

```python
# Load dataset
titanic = sns.load_dataset("titanic")
```

```python
# Basic info
print(titanic.head())
print(titanic.tail())
print(titanic.info())
print(titanic.describe(include="all"))
print("Shape:", titanic.shape)
print("Size:", titanic.size)
print("Dimensions:", titanic.ndim)
print("Columns:", titanic.columns)
```

```python
# 1. Box Plot: Age Distribution by Sex and Survival
plt.figure()
sns.boxplot(x='sex', y='age', hue='survived', data=titanic)
plt.title("Box Plot - Age vs Gender with Survival Info")
plt.legend(title="Survived")
plt.show()
#Females survived more.
#Younger passengers survived more.
#Male age spread is wider.
```

```python

# 2. Count Plot: Number of Passengers by Gender
plt.figure()
sns.countplot(x='sex', data=titanic)
plt.title("Count Plot - Number of Passengers by Gender")
plt.show()#More males than females.
```

```python
# 3. Pie Chart: Survival Ratio survided not survived count
survived_counts = titanic['survived'].value_counts()
plt.figure()
plt.pie(survived_counts, labels=['Not Survived', 'Survived'], autopct='%1.1f%%')
plt.title("Pie Chart - Survival Distribution")
plt.show()
```

```python
# 4. Scatter Plot: Age vs Fare by Survival
plt.figure()
sns.scatterplot(x='age', y='fare', hue='survived', data=titanic)
plt.title("Scatter Plot - Age vs Fare")
plt.show()
#Higher fare passengers survived more.
#Some age groups paid higher fares.
```

```python
# 5. Bar Plot: Average Fare by Passenger Class
plt.figure()
sns.barplot(x='pclass', y='fare', data=titanic)
plt.title("Bar Plot - Average Fare by Passenger Class")
plt.show()#Average fare paid by each passenger class.1st higher 3rd lower
```

```python
# 6. KDE Plot: Age Distribution Kernel Density Estimation.
plt.figure()
sns.kdeplot(titanic['age'].dropna(), fill=True)
plt.title("KDE Plot - Age Distribution")
plt.xlabel("Age")
plt.show()#Continuous age distribution.Most passengers were young adults.
```

```python
# 7. Heatmap: Correlation Matrix Measures relationship between variables.
plt.figure()
corr = titanic[['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare']].corr()
sns.heatmap(corr, annot=True)
plt.title("Heatmap - Correlation Matrix")
plt.show()
#Fare positively related to survival.
#Passenger class negatively related to survival.
```

```python
# 8. Clustermap Groups similar variables together.Groups similar data
sns.clustermap(corr, annot=True)
plt.show()
```
