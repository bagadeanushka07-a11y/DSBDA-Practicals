```python
# Import preprocessing Normalization 
from sklearn import preprocessing
```

```python
# Display dataset
print("\nDataset Preview:")
print(df.head())#Displays updated dataset.
```

```python
# Min-Max Normalization
min_max_scalar = preprocessing.MinMaxScaler()
#Creates normalization object.
x = df.iloc[:, :4]

print("\nOriginal Numerical Data:")
print(x)
```

```python
# Normalize data Converts data into normalized range
x_scaled = min_max_scalar.fit_transform(x)
```

```python
# Create normalized dataframe 
df_normalized = pd.DataFrame(x_scaled)
```

```python

print("\nNormalized Data:")
print(df_normalized)
```

```python
# Import preprocessing
from sklearn import preprocessing
 
```

```python
# Create Label Encoder object Creates object for converting categorical text into numbers.
label_encoder = preprocessing.LabelEncoder()
```

```python
# Unique values before encoding
print("\nUnique Species Before Encoding:")
print(df['species'].unique())
 
```

```python
# Convert categorical values into numerical values
df['species'] = label_encoder.fit_transform(df['species'])
```

```python
# Unique values after encoding
print("\nUnique Species After Encoding:")
print(df['species'].unique())
```
